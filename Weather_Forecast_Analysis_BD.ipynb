## Data Analysis

### Why We Group and Save the Data (Cell 13)
* **Hourly to Daily Data:** The original dataset has 24 hourly rows for every single day (350,880 rows total). After removing the time part, the same `Date` and `City` appear 24 times for each day.
* **Grouping (`groupby` & `mean`):** We group the rows by `Date` and `City` and calculate the daily average (`mean()`) so that each city has **only 1 clean summary record per day** (reducing 350,880 hourly rows to 14,620 daily rows).
* **Sorting & Saving (`sort_values` & `to_csv`):** Finally, we sort the records in calendar order and save them to a new file (`BD_Weather_Cleaned.csv`). This keeps our original dataset safe while giving our application a clean daily CSV file to load, analyze, and update.

```python
# Group hourly rows into daily city averages, sort by Date, and save to a new CSV
df = df.groupby(['Date', 'City'], as_index=False).mean().round(2)
df = df.sort_values(by=['Date', 'City']).reset_index(drop=True)
df.to_csv('BD_Weather_Cleaned.csv', index=False)
df.head(10)
```

Advantage of using Numpy Arrays over list:
* Standard Python Lists do not have built-in .mean(), .median(), or .std() functions.
* NumPy arrays calculate all of these in one simple line and run much faster on large datasets.


```python
# Function to calculate weather statistics for a city using NumPy Arrays
def CalculateCityStats (Dataframe , CityName):
    CityData = Dataframe[Dataframe['City'] == CityName]

    if len(CityData) == 0:
        print("No data found for" , CityName)
        return
    
    TempArray = np.array(CityData['Temp'])
    RainArray = np.array(CityData['Rain'])
    HumidityArray = np.array(CityData['Humidity'])
    SoilArray = np.array(CityData['SoilTemp'])

    print("NumPy Weather Statistics for:" , CityName)
    print("Average Temperature:" , round(np.mean(TempArray),2))
    print("Maximum Temperature:" , round(np.max(TempArray),2))
    print("Minimum Temperature:" , round(np.min(TempArray),2))
    print("Standard Deviation of  Temperature:" , round(np.std(TempArray),2))
    print("Total Rainfall:" , round(np.sum(RainArray),2))
    print("Average Humidity:" , round(np.mean(HumidityArray),2))
    print("Average Soil Temperature:" , round(np.mean(SoilArray),2))
    print()

# Test the Function for City:
CalculateCityStats(df ,'Dhaka')
CalculateCityStats(df ,'Sylhet')

```

## Object Oriented Programming
- Class Representing a single daily weather record
- __ init __ : Stands for initilize. Built in constructor in Python
- self: refres to this object. In C language it is 'this' keyword
- def ToDict(self): A method that packs all the object's stored values into a Python Dictionary. This makes easy for Pandas to convert list of objects into DataFrame or CSV File.

```python
class WeatherRecord:
    # Initializing the object values using the constructor
    def __init__(self, Date, City,Temp,Humidity,Rain,Pressure,CloudCover,WindSpeed, WindDir,SoilTemp):
        self.Date = str(Date)
        self.City = City
        self.RecordId = City + '_' + self.Date
        self.Temp = Temp
        self.Humidity = Humidity
        self.Rain = Rain
        self.Pressure = Pressure
        self.CloudCover = CloudCover
        self.WindSpeed = WindSpeed
        self.WindDir = WindDir
        self.SoilTemp = SoilTemp
        self.Status = self.GetWeatherStatus()

    def GetWeatherStatus(self):
        if self.Rain >= 2.0:
            return "Heavy Rain"
        elif self.Temp >= 32.0:
            return "Heat Wave! Very hot temperature!"
        elif self.WindSpeed >= 25.0:
            return "High Wind!"
        else:
            return "Normal Weather Condition"

    def ToDict(self):
        return{
            'Date' : self.Date,
            'City' : self.City,
            'Temp' : self.Temp,
            'Humidity' : self.Humidity,
            'Rain' : self.Rain,
            'Pressure' : self.Pressure,
            'CloudCover' : self.CloudCover,
            'WindSpeed' : self.WindSpeed,
            'WindDir' : self.WindDir,
            'SoilTemp' : self.SoilTemp,
            'Status' : self.Status
        }
        
```

### User input , Validation and Exception Handling:

```python
def GetValidFloat(UserInput, MinVal, MaxVal):
    while True:
        try:
            Value = float(input(UserInput))
        except ValueError:
            print("Invalid input! Please enter a valid number.")
        else:
            if Value >= MinVal and Value <= MaxVal:
                return Value
            else:
                print("Value must be between", MinVal, "and", MaxVal)
        finally:
            pass

def GetValidCity():
    while True:
        CityInput = input("Enter City (Dhaka, Chittagong, Sylhet, Rajshahi:) ").strip().title()
        if CityInput in AllowedCities:
            return CityInput
        else:
            print("Invalid City! Please choose from:", AllowedCities)

def GetValidDate():
    while True:
        DateInput= input('Enter Date (YYYY-MM-DD): ').strip()
        
        # Check if date has 10 characters and hyphens (-) at index 4 and 7 (YYYY-MM-DD)
        if len(DateInput)==10 and DateInput[4] == '-' and DateInput[7] == '-': 
            return DateInput
        
        else:
            print("Invalid format! Please use YYYY-MM-DD (eg. 2023-05-05)")


```

### Functions to load and save weather records using a CSV file:
- File Functions and Persistent Data

### File Handling Functions: Load and Save Records (Cell 20)
* **`LoadRecordsFromCsv()`:** Reads `BD_Weather_Cleaned.csv` using Pandas inside a `try-except-else-finally` block. It loops through every row, converts each row into a `WeatherRecord` object, and stores all objects inside a Python **List** (`RecordsList`).
* **`SaveRecordsToCsv(RecordsList)`:** Converts all `WeatherRecord` objects back into dictionaries using `.ToDict()`, creates a Pandas DataFrame, and saves the updated data to `BD_Weather_Cleaned.csv` so changes are permanently stored.

```python
def LoadRecordsFromCsv():
    RecordsList = []
    try:
        LoadedDf = pd.read_csv('BD_Weather_Cleaned.csv')
    except FileNotFoundError:
        print("Error: CSV file not found.")
    else:
        # Convert into table rows into a list of dictionaries and create object
        RowList = LoadedDf.to_dict('records')
        for Row in RowList:
            Record = WeatherRecord(
                Row['Date'], Row['City'], Row['Temp'], Row['Humidity'], 
                Row['Rain'], Row['Pressure'], Row['CloudCover'], 
                Row['WindSpeed'], Row['WindDir'], Row['SoilTemp']
            )
            RecordsList.append(Record)
        print("Total records loaded:", len(RecordsList))
    finally:
        print("Total process finished.")
    return RecordsList


# Function to Save records:
def SaveRecordsToCsv(RecordsList):
    try:
        DictList = []
        for Record in RecordsList:
            DictList.append(Record.ToDict())
        SaveDf = pd.DataFrame(DictList)
        SaveDf.to_csv('BD_Weather_Cleaned.csv',index=False)
    except Exception:
        print("Error: Could not sace the file.")
    else:
        print("Total records saved", len(RecordsList))
    finally:
        print("Save Process Finished")
        
```

### Manager Class:
Objectives:

__Part 1__:
- Initialize records list
- View records
- Add new records

__Part 2__:
- Search
- Update
- Delete
- Analyze
- Save Methods

```python
# Part 1:
class WeatherManager:
    def __init__(self):
        #Load all weather objects into python List
        # self.Records = LoadRecordsFromCsv()
        self.Records = LoadRecordsFromCsv()

    def ViewRoceords(self):
        # Convert first 10 objects to dictionaries and display as a Pandas table
        SampleList = []
        for Record in self.Records[:10]:
            SampleList.append(Record.ToDict())
        ViewDf = pd.DataFrame(SampleList)
        print("Showing First 10 Weather Records")
        print(ViewDf)

    def AddRecord(self):
        print("Add New Weather Rocord")
        Date = GetValidDate()
        City = GetValidCity()
        NewId = City + "_" + Date

        # Check if the new RecordId already exsists in the list to prevent duplicates
        for Record in self.Records:
            if Record.RecordId == NewId:
                print("Error: A record for", City, "on", Date, "already exsists!")
                return

        Temp = GetValidFloat("Enter Temperature (-10 to 55): ", -10 ,55)
        Humidity = GetValidFloat("Enter Humidity (0 to 100)", 0, 100)
        Rain = GetValidFloat("Enter Rainfall (0 to 500)", 0, 500)
        Pressure = GetValidFloat("Enter Pressure (900 to 1100)", 900, 1100)
        CloudCover = GetValidFloat("Enter Cloud Cover (0 to 100)", 0, 100)
        WindSpeed = GetValidFloat("Enter Wind Speed (0 to 200)", 0, 200)
        WindDir = GetValidFloat("Enter Wind Direction (0 to 360)", 0, 360)
        SoilTemp = GetValidFloat("Enter Soil Temperature (-5 to 50)", -5, 50)

        NewRecord = WeatherRecord(Date,City,Temp,Humidity,Rain,Pressure,CloudCover,WindSpeed,WindDir,SoilTemp)
        self.Records.append(NewRecord)
        print("Success! New Record has been added for", City, "on" , Date)
        
```

### OOP Principle:
Inheritance (Class inherits attributes and methods from a parent class inside the parentheses)

```python
# Part 2

class WeatherManager(WeatherManager):
    def SearchRecord(self):
        print("Search for Weather Record")
        Date = GetValidDate()
        City = GetValidCity()
        SearchId = City + "_" + Date

        for Record in self.Records:
            if Record.RecordId == SearchId:
                print("Record Found:", Record.ToDict())
                return
        print("Error: No record found for", City, "on", Date)

    def UpdateRecord(self):
        print("Update Weather Record")
        Date = GetValidDate()
        City = GetValidCity()
        TargetId = City + "_" + Date

        for Record in self.Records:
            if Record.RecordId == TargetId:
                Record.Temp = GetValidFloat ("Enter New Temperature (-10 to 55)", -10, 55)
                Record.Humidity = GetValidFloat ("Enter New Humidity (0 to 100)", 0, 100)
                Record.Rain = GetValidFloat ("Enter New Rainfall (0 to 500): ", 0, 500)
                Record.Status = Record.GetWeatherStatus()
                print("Success! Record updated for", City, "on", Date)
                return
        print("Error: No record found to update.")

    def DeleteRecord(self):
        print("Delete Weather Record")
        Date = GetValidDate()
        City = GetValidCity()
        TargetId = City + "_" + Date

        for Record in self.Records:
            if Record.RecordId == TargetId:
                self.Records.remove(Record)
                print("Successfull! Record deleted for", City , "on" , Date)
                return
        print("Error: No record found to delete")

    def AnalyzeCity(self):
        print("City Weather Statistics")
        City = GetValidCity()
        DicList = []
        for Record in self.Records:
            DicList.append(Record.ToDict())
        CurrentDf = pd.DataFrame(DicList)
        CalculateCityStats(CurrentDf, City)

    def SaveData(self):
        SaveRecordsToCsv(self.Records)

```
