## Data Analysis

### Why We Group and Save the Data (Cell 13)
* **Hourly to Daily Data:** The original dataset has 24 hourly rows for every single day (350,880 rows total). After removing the time part, the same `Date` and `City` appear 24 times for each day.
* **Grouping (`groupby` & `mean`):** We group the rows by `Date` and `City` and calculate the daily average (`mean()`) so that each city has **only 1 clean summary record per day** (reducing 350,880 hourly rows to 14,620 daily rows).
* **Sorting & Saving (`sort_values` & `to_csv`):** Finally, we sort the records in calendar order and save them to a new file (`BD_Weather_Cleaned.csv`). This keeps our original dataset safe while giving our application a clean daily CSV file to load, analyze, and update.

```python
# Group hourly rows into daily city averages, sort by Date, and save to a new CSV
df = df.groupby(['Date', 'City'], as_index=False).mean().round(2)
df = df.sort_values(by=['Date', 'City']).reset_index(drop=True)
df.to_csv('BD_Weather_Cleaned.csv', index=False)
df.head(10)
```

## Object Oriented Programming
- Class Representing a single daily weather record
- __ init __ : Stands for initilize. Built in constructor in Python
- self: refres to this object. In C language it is 'this' keyword
- def ToDict(self): A method that packs all the object's stored values into a Python Dictionary. This makes easy for Pandas to convert list of objects into DataFrame or CSV File.

```python
class WeatherRecord:
    # Initializing the object values using the constructor
    def __init__(self, Date, City,Temp,Humidity,Rain,Pressure,CloudCover,WindSpeed, WindDir,SoilTemp):
        self.Date = str(Date)
        self.City = City
        self.RecordId = City + '_' + self.Date
        self.Temp = Temp
        self.Humidity = Humidity
        self.Rain = Rain
        self.Pressure = Pressure
        self.CloudCover = CloudCover
        self.WindSpeed = WindSpeed
        self.WindDir = WindDir
        self.SoilTemp = SoilTemp
        self.Status = self.GetWeatherStatus()

    def GetWeatherStatus(self):
        if self.Rain >= 2.0:
            return "Heavy Rain"
        elif self.Temp >= 32.0:
            return "Heat Wave! Very hot temperature!"
        elif self.WindSpeed >= 25.0:
            return "High Wind!"
        else:
            return "Normal Weather Condition"

    def ToDict(self):
        return{
            'Date' : self.Date,
            'City' : self.City,
            'Temp' : self.Temp,
            'Humidity' : self.Humidity,
            'Rain' : self.Rain,
            'Pressure' : self.Pressure,
            'CloudCover' : self.CloudCover,
            'WindSpeed' : self.WindSpeed,
            'WindDir' : self.WindDir,
            'SoilTemp' : self.SoilTemp,
            'Status' : self.Status
        }
        
```

### User input , Validation and Exception Handling:

```python
def GetValidFloat(UserInput, MinVal, MaxVal):
    while True:
        try:
            Value = float(input(UserInput))
        except ValueError:
            print("Invalid input! Please enter a valid number.")
        else:
            if Value >= MinVal and Value <= MaxVal:
                return Value
            else:
                print("Value must be between", MinVal, "and", MaxVal)
        finally:
            pass

def GetValidCity():
    while True:
        CityInput = input("Enter City (Dhaka, Chittagong, Sylhet, Rajshahi:) ").strip().title()
        if CityInput in AllowedCities:
            return CityInput
        else:
            print("Invalid City! Please choose from:", AllowedCities)

def GetValidDate():
    while True:
        DateInput= input('Enter Date (YYYY-MM-DD): ').strip()
        
        # Check if date has 10 characters and hyphens (-) at index 4 and 7 (YYYY-MM-DD)
        if len(DateInput)==10 and DateInput[4] == '-' and DateInput[7] == '-': 
            return DateInput
        
        else:
            print("Invalid format! Please use YYYY-MM-DD (eg. 2023-05-05)")


```

### Functions to load and save weather records using a CSV file:
- File Functions and Persistent Data

### File Handling Functions: Load and Save Records (Cell 20)
* **`LoadRecordsFromCsv()`:** Reads `BD_Weather_Cleaned.csv` using Pandas inside a `try-except-else-finally` block. It loops through every row, converts each row into a `WeatherRecord` object, and stores all objects inside a Python **List** (`RecordsList`).
* **`SaveRecordsToCsv(RecordsList)`:** Converts all `WeatherRecord` objects back into dictionaries using `.ToDict()`, creates a Pandas DataFrame, and saves the updated data to `BD_Weather_Cleaned.csv` so changes are permanently stored.

```python
def LoadRecordsFromCsv():
    RecordsList = []
    try:
        LoadedDf = pd.read_csv('BD_Weather_Cleaned.csv')
    except FileNotFoundError:
        print("Error: CSV file not found.")
    else:
        # Convert into table rows into a list of dictionaries and create object
        RowList = LoadedDf.to_dict('records')
        for Row in RowList:
            Record = WeatherRecord(
                Row['Date'], Row['City'], Row['Temp'], Row['Humidity'], 
                Row['Rain'], Row['Pressure'], Row['CloudCover'], 
                Row['WindSpeed'], Row['WindDir'], Row['SoilTemp']
            )
            RecordsList.append(Record)
        print("Total records loaded:", len(RecordsList))
    finally:
        print("Total process finished.")
    return RecordsList


# Function to Save records:
def SaveRecordsToCsv(RecordsList):
    try:
        DictList = []
        for Record in RecordsList:
            DictList.append(Record.ToDict())
        SaveDf = pd.DataFrame(DictList)
        SaveDf.to_csv('BD_Weather_Cleaned.csv',index=False)
    except Exception:
        print("Error: Could not sace the file.")
    else:
        print("Total records saved", len(RecordsList))
    finally:
        print("Save Process Finished")
        
```

### Manager Class:
Objectives:

__Part 1__:
- Initialize records list
- View records
- Add new records

__Part 2__:
- Search
- Update
- Delete
- Analyze
- Save Methods

```python
# Part 1:
class WeatherManager:
    def __init__(self):
        #Load all weather objects into python List
        # self.Records = LoadRecordsFromCsv()
        self.Records = LoadRecordsFromCsv()

    def ViewRoceords(self):
        # Convert first 10 objects to dictionaries and display as a Pandas table
        SampleList = []
        for Record in self.Records[:10]:
            SampleList.append(Record.ToDict())
        ViewDf = pd.DataFrame(SampleList)
        print("Showing First 10 Weather Records")
        print(ViewDf)

    def AddRecord(self):
        print("Add New Weather Rocord")
        Date = GetValidDate()
        City = GetValidCity()
        NewId = City + "_" + Date

        # Check if the new RecordId already exsists in the list to prevent duplicates
        for Record in self.Records:
            if Record.RecordId == NewId:
                print("Error: A record for", City, "on", Date, "already exsists!")
                return

        Temp = GetValidFloat("Enter Temperature (-10 to 55): ", -10 ,55)
        Humidity = GetValidFloat("Enter Humidity (0 to 100)", 0, 100)
        Rain = GetValidFloat("Enter Rainfall (0 to 500)", 0, 500)
        Pressure = GetValidFloat("Enter Pressure (900 to 1100)", 900, 1100)
        CloudCover = GetValidFloat("Enter Cloud Cover (0 to 100)", 0, 100)
        WindSpeed = GetValidFloat("Enter Wind Speed (0 to 200)", 0, 200)
        WindDir = GetValidFloat("Enter Wind Direction (0 to 360)", 0, 360)
        SoilTemp = GetValidFloat("Enter Soil Temperature (-5 to 50)", -5, 50)

        NewRecord = WeatherRecord(Date,City,Temp,Humidity,Rain,Pressure,CloudCover,WindSpeed,WindDir,SoilTemp)
        self.Records.append(NewRecord)
        print("Success! New Record has been added for", City, "on" , Date)
        
```

### OOP Principle:
Inheritance (Class inherits attributes and methods from a parent class inside the parentheses)

```python
# Part 2

class WeatherManager(WeatherManager):
    def SearchRecord(self):
        print("Search for Weather Record")
        Date = GetValidDate()
        City = GetValidCity()
        SearchId = City + "_" + Date

        for Record in self.Records:
            if Record.RecordId == SearchId:
                print("Record Found:", Record.ToDict())
                return
        print("Error: No record found for", City, "on", Date)

    def UpdateRecord(self):
        print("Update Weather Record")
        Date = GetValidDate()
        City = GetValidCity()
        TargetId = City + "_" + Date

        for Record in self.Records:
            if Record.RecordId == TargetId:
                Record.Temp = GetValidFloat ("Enter New Temperature (-10 to 55)", -10, 55)
                Record.Humidity = GetValidFloat ("Enter New Humidity (0 to 100)", 0, 100)
                Record.Rain = GetValidFloat ("Enter New Rainfall (0 to 500): ", 0, 500)
                Record.Status = Record.GetWeatherStatus()
                print("Success! Record updated for", City, "on", Date)
                return
        print("Error: No record found to update.")

    def DeleteRecord(self):
        print("Delete Weather Record")
        Date = GetValidDate()
        City = GetValidCity()
        TargetId = City + "_" + Date

        for Record in self.Records:
            if Record.RecordId == TargetId:
                self.Records.remove(Record)
                print("Successfull! Record deleted for", City , "on" , Date)
                return
        print("Error: No record found to delete")

    def AnalyzeCity(self):
        print("City Weather Statistics")
        City = GetValidCity()
        DicList = []
        for Record in self.Records:
            DicList.append(Record.ToDict())
        CurrentDf = pd.DataFrame(DicList)
        CalculateCityStats(CurrentDf, City)

    def SaveData(self):
        SaveRecordsToCsv(self.Records)

```

```python
# WeatherManager, WeatherRecord, AllowedCities, GetValidDate, GetValidCity, GetValidFloat, CalculateCityStats, SaveRecordsToCsv - Done by Shahadat Hossain
import tkinter as tk
from tkinter import ttk, messagebox
import io
import contextlib


class GuiInputFinished(Exception):
    pass


# Bug Fixed: Backend functions read values with input() and show messages with print().
# CallBackend() gives the form values to input() one by one and captures the printed messages,
# so GetValidDate(), GetValidCity() and GetValidFloat() check the GUI values without any change - Done by Shahadat Hossain
def CallBackend(BackendFunction, Answers):
    AnswerList = list(Answers)
    AskedPrompts = []

    def GuiInput(Prompt=""):
        # Backend asks the same question again only when the last value was wrong,
        # so stop there and report the backend error message for that field
        if (AskedPrompts and AskedPrompts[-1] == Prompt) or len(AnswerList) == 0:
            raise GuiInputFinished(Prompt)
        AskedPrompts.append(Prompt)
        return AnswerList.pop(0)

    Output = io.StringIO()
    globals()['input'] = GuiInput
    try:
        with contextlib.redirect_stdout(Output):
            BackendFunction()
    except GuiInputFinished as WrongPrompt:
        Lines = Output.getvalue().strip().splitlines()
        return False, str(WrongPrompt).strip() + "\n" + (Lines[-1] if Lines else "")
    finally:
        del globals()['input']
    return True, Output.getvalue().strip()


def LastMessage(Text):
    Lines = Text.strip().splitlines()
    return Lines[-1] if Lines else ""


# (Label, Field name, Hint) - Field order is the same order the backend AddRecord() asks for the values
# Bug Fixed: Hints show the user the allowed format/range used by GetValidDate() and GetValidFloat() in WeatherManager - Done by Shahadat Hossain
FormFields = [
    ("Date", "Date", "YYYY-MM-DD"),
    ("City", "City", "Select a city"),
    ("Temp (°C)", "Temp", "-10 to 55"),
    ("Humidity (%)", "Humidity", "0 to 100"),
    ("Rain (mm)", "Rain", "0 to 500"),
    ("Pressure (hPa)", "Pressure", "900 to 1100"),
    ("Cloud Cover (%)", "CloudCover", "0 to 100"),
    ("Wind Speed (km/h)", "WindSpeed", "0 to 200"),
    ("Wind Dir (°)", "WindDir", "0 to 360"),
    ("Soil Temp (°C)", "SoilTemp", "-5 to 50")
]

TableColumns = ["Date", "City", "Temp", "Humidity", "Rain", "Pressure",
                "CloudCover", "WindSpeed", "WindDir", "SoilTemp", "Status"]


class WeatherDashboard(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Weather Forecast Analyzer of Bangladesh")
        # Bug Fixed: Window was 1400x820 and did not fit a 15.6 inch laptop screen, now it is smaller and fits the screen - Done by Shahadat Hossain
        ScreenWidth = self.winfo_screenwidth()
        ScreenHeight = self.winfo_screenheight()
        WindowWidth = min(1280, ScreenWidth - 40)
        WindowHeight = min(700, ScreenHeight - 100)
        self.geometry(f"{WindowWidth}x{WindowHeight}+{(ScreenWidth - WindowWidth) // 2}+10")
        self.minsize(min(1000, WindowWidth), min(560, WindowHeight))

        # Backend: WeatherManager() loads every CSV row as a WeatherRecord object using LoadRecordsFromCsv()
        self.Manager = WeatherManager()
        self.FilteredRecords = []

        self.Style = ttk.Style(self)
        self.Style.theme_use("clam")
        self.configure(bg="#eef4ff")
        self.BuildUi()
        self.ApplyFilters()

        # Backend menu option 8: save the data when the window is closed
        self.protocol("WM_DELETE_WINDOW", self.CloseApp)

    def BuildUi(self):
        RootFrame = tk.Frame(self, bg="#eef4ff")
        RootFrame.pack(fill="both", expand=True, padx=18, pady=16)

        Header = tk.Label(RootFrame, text="Weather Forecast Analyzer of Bangladesh",
                          font=("Segoe UI", 22, "bold"), fg="#102a43", bg="#eef4ff")
        Header.pack(anchor="w", pady=(0, 12))

        # Summary Cards Frame
        SummaryFrame = tk.Frame(RootFrame, bg="#eef4ff")
        SummaryFrame.pack(fill="x", pady=(0, 12))

        self.SummaryVars = {
            "total": tk.StringVar(value="0"),
            "city": tk.StringVar(value="0"),
            "avg_temp": tk.StringVar(value="0.0°C"),
            "avg_rain": tk.StringVar(value="0.0mm")
        }
        CardSpecs = [
            ("Total Records", "total", "#dbeafe"),
            ("Cities", "city", "#dcfce7"),
            ("Avg Temp", "avg_temp", "#fef3c7"),
            ("Avg Rain", "avg_rain", "#fce7f3")
        ]
        for Index, (Title, Key, Color) in enumerate(CardSpecs):
            Card = tk.Frame(SummaryFrame, bg=Color, bd=0, highlightthickness=0)
            Card.grid(row=0, column=Index, padx=8, pady=4, sticky="nsew")
            tk.Label(Card, text=Title, bg=Color, fg="#1f2937", font=("Segoe UI", 10, "bold")).pack(pady=(12, 4))
            tk.Label(Card, textvariable=self.SummaryVars[Key], bg=Color, fg="#111827", font=("Segoe UI", 18, "bold")).pack(pady=(0, 12))
            SummaryFrame.grid_columnconfigure(Index, weight=1)

        Content = tk.Frame(RootFrame, bg="#eef4ff")
        Content.pack(fill="both", expand=True)

        # Bug Fixed: Left panel is now scrollable so the Add/Update/Delete buttons can always be reached on a small screen - Done by Shahadat Hossain
        LeftBox = tk.Frame(Content, bg="#ffffff", bd=1, relief="solid")
        LeftBox.pack(side="left", fill="y", padx=(0, 12))
        self.LeftCanvas = tk.Canvas(LeftBox, bg="#ffffff", highlightthickness=0)
        LeftScrollbar = ttk.Scrollbar(LeftBox, orient="vertical", command=self.LeftCanvas.yview)
        self.LeftCanvas.configure(yscrollcommand=LeftScrollbar.set)
        LeftScrollbar.pack(side="right", fill="y")
        self.LeftCanvas.pack(side="left", fill="y")
        Left = tk.Frame(self.LeftCanvas, bg="#ffffff")
        self.LeftCanvas.create_window((0, 0), window=Left, anchor="nw")
        Left.bind("<Configure>", lambda Event: self.LeftCanvas.configure(
            scrollregion=self.LeftCanvas.bbox("all"), width=Left.winfo_reqwidth()))
        self.bind_all("<MouseWheel>", self.OnLeftMouseWheel)

        Right = tk.Frame(Content, bg="#ffffff", bd=1, relief="solid")
        Right.pack(side="left", fill="both", expand=True)

        tk.Label(Left, text="Filters & Record Controls", bg="#ffffff", fg="#0f172a",
                 font=("Segoe UI", 14, "bold")).pack(anchor="w", padx=14, pady=(14, 8))

        # Filters (Backend menu option 3: Search Weather Records)
        FilterFrame = tk.LabelFrame(Left, text="Filter records", bg="#ffffff", fg="#334155",
                                    font=("Segoe UI", 10, "bold"), padx=10, pady=10)
        FilterFrame.pack(fill="x", padx=12, pady=(0, 12))

        tk.Label(FilterFrame, text="City", bg="#ffffff").grid(row=0, column=0, sticky="w", pady=6)
        # Backend: city list comes from the AllowedCities tuple
        self.CityFilter = ttk.Combobox(FilterFrame, values=["All"] + list(AllowedCities), state="readonly", width=26)
        self.CityFilter.grid(row=0, column=1, padx=(8, 0), sticky="ew")
        self.CityFilter.set("All")

        tk.Label(FilterFrame, text="Date", bg="#ffffff").grid(row=1, column=0, sticky="w", pady=6)
        self.DateFilter = ttk.Entry(FilterFrame, width=28)
        self.DateFilter.grid(row=1, column=1, padx=(8, 0), sticky="ew")
        self.DateFilter.insert(0, "YYYY-MM-DD")

        ButtonRow = tk.Frame(FilterFrame, bg="#ffffff")
        ButtonRow.grid(row=2, column=0, columnspan=2, sticky="ew", pady=(10, 0))
        ttk.Button(ButtonRow, text="Apply Filters", command=self.ApplyFilters).pack(side="left", expand=True, padx=(0, 6))
        ttk.Button(ButtonRow, text="Clear", command=self.ClearFilters).pack(side="left", expand=True)

        # Backend menu option 6: City Weather Statistics (By NumPy)
        ttk.Button(FilterFrame, text="City Weather Statistics (NumPy)", command=self.ShowCityStats).grid(
            row=3, column=0, columnspan=2, sticky="ew", pady=(8, 0))

        # Record Form
        FormFrame = tk.LabelFrame(Left, text="Weather record form", bg="#ffffff", fg="#334155",
                                  font=("Segoe UI", 10, "bold"), padx=10, pady=10)
        FormFrame.pack(fill="both", expand=True, padx=12)

        self.RecordFields = {}
        for RowIndex, (LabelText, FieldName, HintText) in enumerate(FormFields):
            tk.Label(FormFrame, text=LabelText, bg="#ffffff").grid(row=RowIndex, column=0, sticky="w", padx=(0, 8), pady=5)

            if FieldName == "City":
                Widget = ttk.Combobox(FormFrame, values=list(AllowedCities), width=16, state="readonly")
            else:
                Widget = ttk.Entry(FormFrame, width=18)
            Widget.grid(row=RowIndex, column=1, sticky="ew", pady=5)
            self.RecordFields[FieldName] = Widget

            tk.Label(FormFrame, text=HintText, bg="#ffffff", fg="#64748b", font=("Segoe UI", 8, "italic")).grid(
                row=RowIndex, column=2, sticky="w", padx=(6, 0))

        ActionFrame = tk.Frame(FormFrame, bg="#ffffff")
        ActionFrame.grid(row=len(FormFields), column=0, columnspan=3, sticky="ew", pady=(10, 0))
        ttk.Button(ActionFrame, text="Add", command=self.AddRecord).pack(side="left", expand=True, padx=2)
        ttk.Button(ActionFrame, text="Update", command=self.UpdateRecord).pack(side="left", expand=True, padx=2)
        ttk.Button(ActionFrame, text="Delete", command=self.DeleteRecord).pack(side="left", expand=True, padx=2)
        ttk.Button(ActionFrame, text="Clear", command=self.ClearForm).pack(side="left", expand=True, padx=2)

        # Table View (Backend menu option 1: View Weather Records)
        TableFrame = tk.Frame(Right, bg="#ffffff")
        TableFrame.pack(fill="both", expand=True, padx=12, pady=12)

        tk.Label(TableFrame, text="Daily Weather Records", bg="#ffffff", fg="#0f172a",
                 font=("Segoe UI", 14, "bold")).pack(anchor="w", pady=(0, 8))

        self.Table = ttk.Treeview(TableFrame, columns=TableColumns, show="headings")
        self.Table.pack(fill="both", expand=True)
        for Column in TableColumns:
            self.Table.heading(Column, text=Column)
            # Bug Fixed: Column width 90 -> 60 so the table fits the smaller window (columns still stretch to fill space) - Done by Shahadat Hossain
            self.Table.column(Column, width=60, anchor="center")
        self.Table.column("Status", width=190, anchor="center")
        self.Table.bind("<<TreeviewSelect>>", self.OnSelectRow)

    # Bug Fixed: Mouse wheel scrolls the left panel when the mouse is over it - Done by Shahadat Hossain
    def OnLeftMouseWheel(self, Event):
        WidgetPath = str(self.tk.call("winfo", "containing", Event.x_root, Event.y_root))
        if WidgetPath.startswith(str(self.LeftCanvas)):
            self.LeftCanvas.yview_scroll(int(-Event.delta / 120), "units")

    def ClearFilters(self):
        self.CityFilter.set("All")
        self.DateFilter.delete(0, "end")
        self.DateFilter.insert(0, "YYYY-MM-DD")
        self.ApplyFilters()

    def ApplyFilters(self):
        City = self.CityFilter.get().strip()
        DateText = self.DateFilter.get().strip()

        if DateText and DateText != "YYYY-MM-DD":
            # Backend: date filter is checked by GetValidDate()
            Ok, Message = CallBackend(GetValidDate, [DateText])
            if not Ok:
                messagebox.showerror("Invalid Date", Message)
                self.DateFilter.delete(0, "end")
                self.DateFilter.insert(0, "YYYY-MM-DD")
                DateText = ""
        else:
            DateText = ""

        # Search the WeatherRecord objects stored in the Manager's list
        Matches = []
        for Record in self.Manager.Records:
            if City != "All" and Record.City != City:
                continue
            if DateText and Record.Date != DateText:
                continue
            Matches.append(Record)

        # Newest records (and newly added ones) are shown first
        self.FilteredRecords = Matches[::-1]
        self.RefreshDashboard()
        self.RefreshTable()

    def RefreshDashboard(self):
        if len(self.FilteredRecords) == 0:
            self.SummaryVars["total"].set("0")
            self.SummaryVars["city"].set("0")
            self.SummaryVars["avg_temp"].set("0.0°C")
            self.SummaryVars["avg_rain"].set("0.0mm")
            return

        # NumPy arrays for the summary cards, same idea as CalculateCityStats()
        TempArray = np.array([Record.Temp for Record in self.FilteredRecords])
        RainArray = np.array([Record.Rain for Record in self.FilteredRecords])
        CitySet = set(Record.City for Record in self.FilteredRecords)

        self.SummaryVars["total"].set(str(len(self.FilteredRecords)))
        self.SummaryVars["city"].set(str(len(CitySet)))
        self.SummaryVars["avg_temp"].set(f"{np.mean(TempArray):.1f}°C")
        self.SummaryVars["avg_rain"].set(f"{np.mean(RainArray):.1f}mm")

    def RefreshTable(self):
        self.Table.delete(*self.Table.get_children())
        for Position, Record in enumerate(self.FilteredRecords):
            # Backend: WeatherRecord.ToDict() gives the values (Status comes from GetWeatherStatus())
            RecordDict = Record.ToDict()
            self.Table.insert("", "end", iid=str(Position), values=[RecordDict[Column] for Column in TableColumns])

    def OnSelectRow(self, Event):
        Selected = self.Table.selection()
        if not Selected:
            return
        Record = self.FilteredRecords[int(Selected[0])]
        self.FillForm(Record.ToDict())

    def FillForm(self, RecordDict):
        for Key, Widget in self.RecordFields.items():
            if isinstance(Widget, ttk.Combobox):
                Widget.set(str(RecordDict.get(Key, "")))
            else:
                Widget.delete(0, "end")
                Widget.insert(0, str(RecordDict.get(Key, "")))

    def ClearForm(self):
        for Widget in self.RecordFields.values():
            if isinstance(Widget, ttk.Combobox):
                Widget.set("")
            else:
                Widget.delete(0, "end")

    def GetFormValues(self):
        Values = {}
        for Key, Widget in self.RecordFields.items():
            Values[Key] = Widget.get().strip()
        return Values

    def SaveAndRefresh(self):
        # Backend menu option 7: WeatherManager.SaveData() -> SaveRecordsToCsv()
        Ok, Message = CallBackend(self.Manager.SaveData, [])
        self.ApplyFilters()
        self.ClearForm()
        if "Error" in Message:
            messagebox.showerror("Save Error", Message)
            return False
        return True

    def AddRecord(self):
        # Backend menu option 2: WeatherManager.AddRecord() checks date, city, duplicates and every value range
        Values = self.GetFormValues()
        Answers = [Values[FieldName] for (LabelText, FieldName, HintText) in FormFields]
        CountBefore = len(self.Manager.Records)

        Ok, Message = CallBackend(self.Manager.AddRecord, Answers)
        if not Ok:
            messagebox.showerror("Validation Error", Message)
        elif len(self.Manager.Records) == CountBefore:
            messagebox.showerror("Duplicate Record", LastMessage(Message))
        elif self.SaveAndRefresh():
            messagebox.showinfo("Success", LastMessage(Message))

    def UpdateRecord(self):
        # Backend menu option 4: WeatherManager.UpdateRecord() finds the record by Date + City
        # and updates Temp, Humidity and Rain
        Values = self.GetFormValues()
        TargetId = Values["City"].title() + "_" + Values["Date"]

        # Keep the old values: backend UpdateRecord() changes Temp before it checks Humidity and Rain,
        # so if a later value is wrong the record is put back the way it was
        Backup = None
        for Record in self.Manager.Records:
            if Record.RecordId == TargetId:
                Backup = (Record, Record.Temp, Record.Humidity, Record.Rain, Record.Status)
                break

        Answers = [Values["Date"], Values["City"], Values["Temp"], Values["Humidity"], Values["Rain"]]
        Ok, Message = CallBackend(self.Manager.UpdateRecord, Answers)
        if not Ok:
            if Backup is not None:
                Record = Backup[0]
                Record.Temp, Record.Humidity, Record.Rain, Record.Status = Backup[1:]
            messagebox.showerror("Validation Error", Message)
        elif "Success" not in Message:
            messagebox.showerror("Update Record", LastMessage(Message))
        elif self.SaveAndRefresh():
            messagebox.showinfo("Success", LastMessage(Message) + "\n(Temp, Humidity and Rain were updated)")

    def DeleteRecord(self):
        # Backend menu option 5: WeatherManager.DeleteRecord() finds the record by Date + City
        Values = self.GetFormValues()
        if not messagebox.askyesno("Delete record", "Are you sure you want to delete this record?"):
            return

        CountBefore = len(self.Manager.Records)
        Ok, Message = CallBackend(self.Manager.DeleteRecord, [Values["Date"], Values["City"]])
        if not Ok:
            messagebox.showerror("Validation Error", Message)
        elif len(self.Manager.Records) == CountBefore:
            messagebox.showerror("Delete Record", LastMessage(Message))
        elif self.SaveAndRefresh():
            messagebox.showinfo("Success", LastMessage(Message))

    def ShowCityStats(self):
        # Backend menu option 6: WeatherManager.AnalyzeCity() -> GetValidCity() + CalculateCityStats()
        Ok, Message = CallBackend(self.Manager.AnalyzeCity, [self.CityFilter.get()])
        if not Ok:
            messagebox.showwarning("Select a city", Message)
        else:
            messagebox.showinfo("City Weather Statistics", Message)

    def CloseApp(self):
        # Backend menu option 8: save data and close the program
        Ok, Message = CallBackend(self.Manager.SaveData, [])
        if "Error" in Message:
            messagebox.showerror("Save Error", Message)
        self.destroy()


# Run the GUI Dashboard (replaces the old terminal RunApplication() call)
App = WeatherDashboard()
App.mainloop()
```
